## Silver Table Example

```python
%sql

CREATE OR REFRESH STREAMING LIVE TABLE pitching_year_numbers(
  CONSTRAINT valid_ip EXPECT (ip > 0) ON VIOLATION DROP ROW
)

COMMENT "Roll up Stats for a pitcher up the the yearly level"
as
SELECT player, name, year, sum(wins) as wins,
sum(losses) as losses, sum(ip) as ip,
sum(so) as so, sum(bb) as bb,
sum(hbp) as hbp, sum(sv) as sv,
sum(er) as er, sum(war) as war
FROM LIVE.pitching_raw
where game_type = 'R'
group by 1,2,3
order by year desc, sum(war) desc;
```
